# Predicción de Readmisión Hospitalaria en Pacientes Diabéticos: Revisión del Estado dela literatura y Propuesta de un Nuevo Enfoque

## 1. Introducción y Contexto Clínico

La diabetes representa uno de los desafíos más significativos para los sistemas de salud modernos. A pesar de la existencia de evidencia de alta calidad que demuestra que las intervenciones preventivas y terapéuticas mejoran los resultados clínicos, una proporción alarmante de pacientes no recibe el cuidado adecuado.

Esta brecha en la atención se atribuye parcialmente a una gestión arbitraria de la diabetes en entornos hospitalarios, donde el control glucémico a menudo no recibe la prioridad necesaria. La consecuencia directa es el aumento de las **readmisiones hospitalarias tempranas** (dentro de los 30 días posteriores al alta).

El problema de la readmisión no es solo financiero —incrementando los costos operativos para hospitales y redes de salud— sino fundamentalmente humano, impactando la morbilidad y mortalidad de pacientes que enfrentan complicaciones severas.

## 2. Definición del Problema

El objetivo central de este estudio se basa en un conjunto de datos históricos que abarca diez años (1999-2008) de atención clínica en 130 hospitales y redes de entrega integrada en EE.UU. Cada registro representa la estancia hospitalaria de un paciente diagnosticado con diabetes, incluyendo pruebas de laboratorio, medicamentos y estancias de hasta 14 días.

La tarea analítica es **determinar la probabilidad de readmisión temprana del paciente ( < 30 días)** basándonos en estas variables clínicas y administrativas.

## 4. Revisión de la Literatura

A continuación, se presentan los hallazgos de investigaciones previas que han abordado la problemática de la readmisión utilizando el mismo conjunto de datos o contextos clínicos similares.

### Paper 1: Determinants and predictors of early re-admission of patients with hyperglycemic crises

**Resumen del Enfoque:**
Este estudio se centró en pacientes admitidos por crisis hiperglucémicas utilizando el dataset público de hospitales de EE.UU. (1999-2008). El objetivo principal fue superar las limitaciones de las técnicas estadísticas clásicas mediante el uso de Inteligencia Artificial para identificar pacientes en riesgo.

**Metodología y Modelo:**
Los autores evaluaron inicialmente tres modelos de Machine Learning, seleccionando aquel que maximizara la capacidad de detectar los casos positivos (readmisión).
* **Modelo Seleccionado:** Extreme Gradient Boosting (XGBoost).
* **Variables Importantes:** El modelo identificó la edad, el número de medicamentos, las visitas previas a emergencias/hospitalización y los procedimientos no de laboratorio como los predictores más fuertes.

**Resultados y Métricas Clave:**
El estudio reportó una prevalencia de readmisión del **13.32%**. El modelo XGBoost obtuvo los siguientes resultados de desempeño:
* **Recall (Sensibilidad) para la clase positiva:** 79%
* **Área bajo la curva (AUC-ROC):** 0.78

**Conclusión Relevante para nuestro estudio:**
Un hallazgo crucial de este paper es la demostración de la superioridad del Machine Learning en el ámbito clínico frente a la estadística tradicional. Mientras que las pruebas estadísticas clásicas (como Chi-cuadrado) mostraron correlaciones débiles entre la readmisión y variables demográficas (raza, sexo, edad), el modelo de XGBoost logró capturar patrones complejos no lineales, validando la utilidad de estas herramientas para la toma de decisiones en la gestión de la diabetes.

### Paper 2: Comparison of machine learning models for predicting 30-day readmission rates

**Resumen del Enfoque:**
A diferencia de estudios que se centran en un único algoritmo, esta investigación realizó una comparativa exhaustiva entre **11 métodos distintos**, contrastando modelos de Machine Learning tradicional (como SVM, Decision Trees, RF) contra modelos de Deep Learning (específicamente LSTM - Long Short-Term Memory). El objetivo fue predecir la readmisión a 30 días utilizando el dataset "Diabetes 130-US Hospitals" (UCI Repository).

**Factor Diferencial: Optimización "Grey Wolf" (GWO)**
El aspecto más innovador de este estudio fue la incorporación del **Grey Wolf Optimizer (GWO)** para la selección de características.
* *¿Qué es GWO?*: Es un algoritmo de optimización basado en "inteligencia de enjambre" (swarm intelligence) que imita la jerarquía de liderazgo y los mecanismos de caza de los lobos grises en la naturaleza.
* *Su uso:* Los autores utilizaron GWO para seleccionar de manera inteligente las variables más relevantes del dataset antes de entrenar los modelos, combinándolo con técnicas de balanceo de datos (SMOTE) para manejar la desproporción de clases.

**Resultados y Métricas Clave:**
Contrario a la creencia popular de que el Deep Learning siempre es superior, este estudio encontró que los algoritmos de ensamblaje tradicionales, potenciados con GWO, ofrecieron mejor rendimiento:

* **Modelo Ganador:** **Random Forest (RF)**. Emergió como el modelo más favorable, equilibrando mejor la precisión y el recall.
    * **Accuracy:** 0.88
    * **F1-Score:** 0.83
* **Segundo Lugar:** XGBoost mostró resultados muy competitivos y cercanos al RF (Accuracy: 0.88, F1: 0.84*).
* **Conclusión Comparativa:** Los autores concluyen que los modelos RF y XGBoost con optimización GWO superaron al modelado predictivo de Deep Learning (LSTM) en escenarios de readmisión por diabetes.

> **Nota:** Aunque XGBoost obtuvo un F1-Score marginalmente más alto en los reportes brutos, los autores determinaron que Random Forest fue el modelo más consistente y favorable en la evaluación global de métricas.

### Paper 3: Implementation of machine learning algorithms to create diabetic patient re-admission profiles

**Resumen del Enfoque:**
El objetivo distintivo de este estudio no fue únicamente la predicción binaria, sino el **reconocimiento de patrones** para construir "perfiles" de readmisión. Los autores buscaron combinaciones de factores que caracterizan o explican *quiénes* son los pacientes que regresan al hospital, moviéndose de la predicción pura a la descripción del fenómeno.

**Novedad Algorítmica:**
Este estudio exploró un conjunto de clasificadores diferentes a los vistos en los papers anteriores (que se centraban en Boosting o Deep Learning), introduciendo algoritmos más interpretables o basados en estadística clásica y distancia:
* **Linear Discriminant Analysis (LDA)**
* **k-Nearest Neighbor (k-NN)**
* **Naive Bayes**
* **J48** (Una implementación popular del algoritmo de árbol de decisión C4.5)

**Construcción de Clases y Hallazgos:**
La metodología se centró en analizar las "clases" o reglas producidas por estos modelos para segmentar a la población. De un total de 100,000 casos (con una tasa de readmisión del 47%), los algoritmos permitieron perfilar al paciente típico de alto riesgo.

* **El Perfil de Riesgo:** Los modelos indicaron que la readmisión es más probable en pacientes mujeres, caucásicos, o pacientes ambulatorios.
* **El Factor "Cuidado Insuficiente":** Un hallazgo crítico derivado de este perfilamiento es que la readmisión está fuertemente vinculada a la **falta de rigor en el tratamiento inicial**. Los pacientes que recibieron menos procedimientos de laboratorio, menos medicación, o que fueron dados de alta sin administración de insulina (a pesar de tener HbA1c positivo) mostraron mayor tendencia a regresar.

**Conclusión Relevante:**
Este paper demuestra que los algoritmos de Machine Learning (como J48 o Naive Bayes) pueden actuar como herramientas de auditoría clínica, revelando que los pacientes que no reciben una evaluación vigorosa o mejoras claras antes del alta son los más vulnerables.

## 5. Propuesta de Valor y Estrategia de Modelado: Stacking Ensemble

### Identificación de la Oportunidad
Tras la revisión exhaustiva de la literatura, hemos identificado un patrón recurrente: los estudios previos tienden a enfocar el problema como una **competencia** entre algoritmos.
* El **Paper 1** demostró la potencia del *Gradient Boosting* (XGBoost).
* El **Paper 2** evidenció la robustez del *Random Forest* optimizado.
* El **Paper 3** resaltó cómo diferentes clasificadores capturan distintos perfiles de pacientes.

Sin embargo, ninguno de estos estudios exploró la posibilidad de la **cooperación algorítmica**. Existe un vacío en la literatura consultada respecto al uso de arquitecturas de ensamblaje avanzado (Stacking) para este dataset específico.

### Nuestra Propuesta: Arquitectura "Stacking Classifier"

Para este proyecto, proponemos la implementación de un **Stacking Ensemble Classifier**. En lugar de seleccionar un único "mejor modelo", construiremos una arquitectura en dos niveles:

1.  **Nivel 0 (Base Learners):** Entrenaremos modelos fuertes y diversos (heterogéneos) en paralelo. Específicamente, utilizaremos los campeones de los papers anteriores: **XGBoost** (por su manejo de sesgo) y **Random Forest** (por su manejo de varianza), sumando posiblemente un modelo lineal o basado en distancia para diversidad.
2.  **Nivel 1 (Meta-Learner):** Las predicciones de los modelos base servirán como "nuevos datos" (inputs) para un meta-modelo final (generalmente una Regresión Logística o un modelo lineal simple), que aprenderá a ponderar qué modelo base tiene la razón en cada tipo de caso.



### Beneficios Esperados
La hipótesis central de esta propuesta es que el Stacking permitirá:
* **Reducir el error de generalización:** Al combinar modelos con diferentes fundamentos matemáticos (Bagging vs. Boosting), el sistema será más robusto ante el ruido.
* **Superar el Techo de Desempeño:** Esperamos obtener métricas (AUC-ROC y F1-Score) superiores a las reportadas individualmente en la literatura (superando el 0.78 AUC del Paper 1 y el 0.83 F1 del Paper 2).
* **Estabilidad:** Un modelo apilado suele ser menos propenso a fluctuaciones drásticas cuando se enfrenta a datos nuevos, ofreciendo una solución más segura para un entorno clínico real.
